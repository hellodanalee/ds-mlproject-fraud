### 3. Key Decision: Use of `imbalanced-learn` for Imbalance Handling ✅

**Why?**  
Imbalanced datasets (like fraud detection) require special treatment. We chose the [`imbalanced-learn`](https://imbalanced-learn.org/) library to address this issue because it:

- Integrates seamlessly with `scikit-learn` Pipelines  
- Provides robust oversampling/undersampling techniques  
- Ensures reproducibility and modular experimentation  
- Allows us to test different strategies in a clean, unified way

**Sampling techniques we compared:**
- `SMOTE` — Synthetic Minority Oversampling Technique  
- `ADASYN` — Adaptive Synthetic Sampling  
- `SMOTE + Tomek Links` — Oversampling + data cleaning  
- `scale_pos_weight` — Native class-weight adjustment in XGBoost

Each strategy was plugged into a pipeline and compared using cross-validation.

---

### 4. Hyperparameter Tuning  
- `RandomizedSearchCV` (50 iterations)  
- Stratified 5-Fold Cross-Validation  
- Custom F1 scorer for fraud class (`pos_label=1`)

---

### 5. Threshold Optimization  
- Precision-Recall Curve analysis on training set  
- Selected optimal classification threshold (`best_threshold`)  
- Used instead of default 0.5 to **maximize F1-score**

💡 The F1-score balances **Recall** (fraud detection rate) and **Precision** (false positive control),  
making it ideal for evaluating imbalanced problems like fraud detection.

---

## 📦 Sampling Strategy Summary

| Sampling Method      | Description                                                                                     |
|----------------------|-------------------------------------------------------------------------------------------------|
| **SMOTE**            | Generates synthetic examples of the minority class (fraud) to help balance the training data.   |
| **ADASYN**           | Like SMOTE, but focuses more on hard-to-classify fraud examples by generating more near them.   |
| **SMOTETomek**       | Combines SMOTE with a cleaning step that removes overlapping or ambiguous samples (Tomek Links). |
| **scale_pos_weight** | Does not modify the data. Instead, it tells the model to give more weight to rare fraud cases.  |

---

### 💡 Key Insight

All techniques aim to solve the class imbalance problem in fraud detection.  
However, **`scale_pos_weight`** is:

- The simplest to apply  
- Computationally efficient  
- The best-performing method in our evaluation

Use this setting when working with **XGBoost** and **highly imbalanced data**.

## 📐 Classification Metrics: Plaintext Formulas

Let:

- **TP** = True Positives  
- **FP** = False Positives  
- **FN** = False Negatives  

---

### 🎯 Precision

Measures how many of the predicted positive cases were actually correct.

**Precision = TP / (TP + FP)**

---

### 🔍 Recall

Measures how many of the actual positive cases were correctly identified.

**Recall = TP / (TP + FN)**

---

### ⚖️ F1-Score

The harmonic mean of Precision and Recall.

**F1 = 2 × (Precision × Recall) / (Precision + Recall)**

---

### 💡 Interpretation

- **High Precision** → Few false alarms  
- **High Recall** → Few missed fraud cases  
- **High F1** → Balanced performance

````python
def compare_df_structure(df1: pd.DataFrame, df2: pd.DataFrame):
    differences = []
    
    # --- Spaltennamen ---
    cols1 = list(df1.columns)
    cols2 = list(df2.columns)
    
    if cols1 != cols2:
        differences.append("⚠️ Spalten unterschiedlich in Reihenfolge oder Namen:")
        # Welche fehlen in df2?
        missing_in_df2 = [c for c in cols1 if c not in cols2]
        if missing_in_df2:
            differences.append(f"  In df1, nicht in df2: {missing_in_df2}")
        # Welche fehlen in df1?
        missing_in_df1 = [c for c in cols2 if c not in cols1]
        if missing_in_df1:
            differences.append(f"  In df2, nicht in df1: {missing_in_df1}")
        # Gleiche Namen, aber andere Reihenfolge
        if set(cols1) == set(cols2) and cols1 != cols2:
            differences.append("  Gleiche Spalten, aber Reihenfolge unterschiedlich")
    else:
        differences.append("✅ Spaltennamen und Reihenfolge identisch")
    
    # --- Index / Zeilenreihenfolge ---
    idx1 = list(df1.index)
    idx2 = list(df2.index)
    
    if idx1 != idx2:
        differences.append("⚠️ Index-Reihenfolge unterschiedlich")
        # Fehlende Einträge:
        missing_idx_in_df2 = [i for i in idx1 if i not in idx2]
        if missing_idx_in_df2:
            differences.append(f"  In df1, nicht in df2: {missing_idx_in_df2[:10]}{'...' if len(missing_idx_in_df2) > 10 else ''}")
        missing_idx_in_df1 = [i for i in idx2 if i not in idx1]
        if missing_idx_in_df1:
            differences.append(f"  In df2, nicht in df1: {missing_idx_in_df1[:10]}{'...' if len(missing_idx_in_df1) > 10 else ''}")
    else:
        differences.append("✅ Index-Reihenfolge identisch")
    
    return "\n".join(differences)

# Beispiel
# print(compare_df_structure(df1, df2))




def diff_column_positions(df_train: pd.DataFrame, df_test: pd.DataFrame, limit=30):
    cols_tr = list(df_train.columns)
    cols_te = list(df_test.columns)
    out = []

    if set(cols_tr) != set(cols_te):
        only_tr = [c for c in cols_tr if c not in cols_te]
        only_te = [c for c in cols_te if c not in cols_tr]
        out.append(f"⚠️ Unterschiedliche Spaltenmengen.")
        if only_tr: out.append(f"  Nur in Train: {only_tr[:limit]}{' ...' if len(only_tr)>limit else ''}")
        if only_te: out.append(f"  Nur in Test : {only_te[:limit]}{' ...' if len(only_te)>limit else ''}")
        return "\n".join(out)

    # gleiche Menge, Reihenfolge prüfen
    mismatches = [(i, cols_tr[i], cols_te[i]) for i in range(len(cols_tr)) if cols_tr[i] != cols_te[i]]
    if not mismatches:
        return "✅ Spaltenreihenfolge identisch."
    out.append(f"⚠️ Reihenfolge unterscheidet sich an {len(mismatches)} Position(en). Beispiele:")
    for i, a, b in mismatches[:limit]:
        out.append(f"  Pos {i}: Train='{a}' | Test='{b}'")
    return "\n".join(out)
````

````python
# --- Nur scale_pos_weight Variante mit Multi-Metric-Optimierung ---

from imblearn.pipeline import Pipeline
from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV, cross_validate
from sklearn.metrics import make_scorer, f1_score, precision_score, recall_score, \
                            average_precision_score, roc_auc_score, balanced_accuracy_score, \
                            log_loss, matthews_corrcoef, precision_recall_curve, PrecisionRecallDisplay
from scipy.stats import randint, uniform
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- Data ---
X = df_fraud_aggregated.drop(columns=["target", "client_id"]).fillna(0)
y = df_fraud_aggregated["target"].astype(int)
X_test = df_fraud_aggregated_test_merged.drop(columns=["client_id"]).fillna(0)

print(compare_df_structure(X, X_test))
print(diff_column_positions(X, X_test))

# --- CV ---
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# --- Multi-Metric Scoring ---
scoring = {
    "f1": make_scorer(f1_score, pos_label=1, zero_division=0),
    "roc_auc": "roc_auc",
    "avg_precision": "average_precision"
}

# --- Refit-Regel: gewichtete Kombination ---
w_f1, w_roc, w_pr = 0.6, 0.3, 0.1
def refit_by_combo(cv_results):
    f1  = np.array(cv_results["mean_test_f1"])
    roc = np.array(cv_results["mean_test_roc_auc"])
    pr  = np.array(cv_results["mean_test_avg_precision"])
    combo = w_f1 * f1 + w_roc * roc + w_pr * pr
    return int(np.nanargmax(combo))

# --- Class imbalance ratio ---
ratio = (y == 0).sum() / max((y == 1).sum(), 1)

# --- Pipeline ---
xgb = XGBClassifier(
    eval_metric="logloss",
    scale_pos_weight=ratio,
    random_state=42,
    verbosity=0,
    n_jobs=-1,
    tree_method="hist"
)
pipeline = Pipeline([("xgb", xgb)])

# --- Hyperparameter-Suche ---
param_dist = {
    "xgb__n_estimators": randint(100, 500),
    "xgb__max_depth": randint(3, 10),
    "xgb__learning_rate": uniform(0.01, 0.3),
    "xgb__subsample": uniform(0.5, 0.5),
    "xgb__colsample_bytree": uniform(0.5, 0.5),
    "xgb__min_child_weight": randint(1, 10),
    "xgb__gamma": uniform(0, 5),
    "xgb__scale_pos_weight": [ratio]
}

search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_dist,
    n_iter=30,
    scoring=scoring,
    refit=refit_by_combo,
    cv=cv,
    verbose=0,
    n_jobs=-1,
    random_state=42
)

search.fit(X, y)
best_model = search.best_estimator_

cvres = search.cv_results_
print(
    "Best (combo) params:", search.best_params_, "\n",
    f"mean F1={cvres['mean_test_f1'][search.best_index_]:.4f}, ",
    f"ROC-AUC={cvres['mean_test_roc_auc'][search.best_index_]:.4f}, ",
    f"PR-AUC={cvres['mean_test_avg_precision'][search.best_index_]:.4f}"
)

# ======================================================================
# --- Threshold-Bestimmung: PRC-basiert + feiner Scan + lokaler Zoom ---
# ======================================================================

def f1_at_threshold(y_true, y_score, thr):
    preds = (y_score >= thr).astype(int)
    return f1_score(y_true, preds, zero_division=0)

y_proba_train = best_model.predict_proba(X)[:, 1]

# 1) PRC-basierte Kandidaten (diskret)
precisions, recalls, thresholds_pr = precision_recall_curve(y, y_proba_train)
f1_pr = 2 * (precisions[:-1] * recalls[:-1]) / (precisions[:-1] + recalls[:-1] + 1e-10)
idx_pr = int(np.argmax(f1_pr))
thr_pr = float(thresholds_pr[idx_pr])
f1_best_pr = float(f1_pr[idx_pr])

# 2) Globaler feiner Scan
n_fine = 2001
thr_fine_grid = np.linspace(0.0, 1.0, n_fine)
f1_fine = np.array([f1_at_threshold(y, y_proba_train, thr) for thr in thr_fine_grid])
idx_fine = int(np.argmax(f1_fine))
thr_fine = float(thr_fine_grid[idx_fine])
f1_best_fine = float(f1_fine[idx_fine])

# 3) Lokaler Zoom
thr_seed = thr_fine if f1_best_fine >= f1_best_pr else thr_pr
zoom_width = 0.05
lo = max(0.0, thr_seed - zoom_width)
hi = min(1.0, thr_seed + zoom_width)
n_zoom = 2001
thr_zoom_grid = np.linspace(lo, hi, n_zoom)
f1_zoom = np.array([f1_at_threshold(y, y_proba_train, thr) for thr in thr_zoom_grid])
idx_zoom = int(np.argmax(f1_zoom))
thr_zoom = float(thr_zoom_grid[idx_zoom])
f1_best_zoom = float(f1_zoom[idx_zoom])

# 4) Finale Auswahl
candidates = [
    ("PRC",  thr_pr,   f1_best_pr),
    ("FINE", thr_fine, f1_best_fine),
    ("ZOOM", thr_zoom, f1_best_zoom),
]
best_name, best_threshold, best_f1 = max(candidates, key=lambda t: t[2])

print(f"Best threshold method : {best_name}")
print(f"Best threshold value  : {best_threshold:.4f}")
print(f"Best F1 (Train)       : {best_f1:.4f}")

y_pred_train = (y_proba_train >= best_threshold).astype(int)
print(f"Precision@thr         : {precision_score(y, y_pred_train, zero_division=0):.4f}")
print(f"Recall@thr            : {recall_score(y, y_pred_train, zero_division=0):.4f}")

# --- PR-Kurve plotten ---
plt.figure()
PrecisionRecallDisplay(precision=precisions, recall=recalls).plot()
plt.axvline(recalls[idx_pr], linestyle="--")  # PRC-Bestpunkt
recall_final = recall_score(y, y_pred_train, zero_division=0)
plt.axvline(recall_final, linestyle=":")      # final gewählter Threshold
plt.title("PR Curve - scale_pos_weight (--- PRC, : final best)")
plt.show()

# --- Finale CV-Evaluation ---
scoring_eval = {
    "accuracy": "accuracy",
    "balanced_acc": "balanced_accuracy",
    "precision": make_scorer(precision_score, pos_label=1, zero_division=0),
    "recall": make_scorer(recall_score, pos_label=1, zero_division=0),
    "f1": make_scorer(f1_score, pos_label=1, zero_division=0),
    "roc_auc": "roc_auc",
    "avg_precision": "average_precision",
    "log_loss": "neg_log_loss",
    "mcc": make_scorer(matthews_corrcoef)
}
cv_result = cross_validate(
    best_model, X, y,
    cv=cv,
    scoring=scoring_eval,
    return_train_score=False,
    n_jobs=-1
)
print("\n=== scale_pos_weight Evaluation Summary ===")
for metric, values in cv_result.items():
    if metric.startswith("test_"):
        print(f"{metric[5:]:<15}: {np.mean(values):.3f} ± {np.std(values):.3f}")
print(f"Optimal Threshold (Train): {best_threshold:.3f}")

# --- Submission ---
y_test_proba = best_model.predict_proba(X_test)[:, 1]
y_test_pred = (y_test_proba >= best_threshold).astype(int)

df_submission = df_fraud_aggregated_test_merged[['client_id']].copy()
df_submission["target"] = y_test_proba  # oder y_test_pred
filename = f"fraud_predictions__combi_zindi_scale_pos_weight_thr{best_threshold:.3f}.csv"
df_submission.to_csv(filename, index=False, float_format='%.16f')
print(f"Predictions saved to: {filename}")
````
