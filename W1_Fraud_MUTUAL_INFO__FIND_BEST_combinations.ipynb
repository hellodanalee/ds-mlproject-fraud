```python
from fraud import collectAllFeaturesBaseline, collectAllFeaturesBaselineTest
from fraud import calculate_mutual_information, visualize_mutual_information, addCombinations
from fraudfeaturefinder import fraudfeaturefinder, apply_saved_combinations
```

```python
df_features = collectAllFeaturesBaseline(without_t_features=True)
df_features
```

```python
df_features_mutual = calculate_mutual_information(df_features, target_col="target", exclude_cols=["client_id", "target"])
df_features_mutual.head(100)
```

```python
top_n_features = df_features_mutual.head(20)["Feature"].tolist()
print(top_n_features)
```

```python
# 2. Liste der Spalten, die ignoriert werden sollen (z. B. ID-Spalten oder Kategorien)
ignore_cols = ['client_id', 'target']

# 3. Initialisiere das Objekt
finder = fraudfeaturefinder(df_feature=df_features, target_col='target', threshold=0.025, ignore_cols=ignore_cols, operations=["*", "/"], selected_features=top_n_features)

# 4. Suche nach guten Feature-Kombinationen
finder.find_good_combinations()

# 5. Ergebnisse speichern
finder.save_results("./data/mutual-info-combinations.csv")
```
